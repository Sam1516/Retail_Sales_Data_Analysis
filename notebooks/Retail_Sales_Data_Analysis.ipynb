# Retail Sales Data Exploration & Analysis

## Project Overview

This project focuses on exploring and analyzing retail sales data to identify sales trends, profitability patterns, top-performing products, customer performance, regional performance, and the impact of discounts on profit.

## Objectives

- Analyze overall sales and profit performance
- Identify monthly and yearly sales trends
- Analyze category and sub-category performance
- Identify top-selling and most profitable products
- Compare regional sales performance
- Analyze customer profitability
- Study the relationship between discount and profit
- Generate meaningful business insights

# Dataset Information

The dataset contains retail sales transaction data.

### Key Columns

- `order_date` — Date of the order
- `sales` — Sales amount
- `quantity` — Quantity of products sold
- `discount` — Discount applied
- `profit` — Profit generated
- `shipping_cost` — Shipping cost
- `category` — Product category
- `sub_category` — Product sub-category
- `product_name` — Product name
- `customer_name` — Customer name
- `region` — Sales region
- `segment` — Customer segment
- `ship_mode` — Shipping method

### Dataset Size

- **Rows:** 51,290
- **Numerical analysis columns:** Sales, Quantity, Discount, Profit, Shipping Cost

# Final Business Insights

- The business shows strong overall growth from 2011 to 2014.
- Sales increased from 2.26M in 2011 to 4.30M in 2014.
- 2013 recorded the highest year-over-year growth in both sales and profit.
- Technology is the strongest category in terms of sales, profit, and profit margin.
- Furniture has relatively low profitability despite having high sales.
- Tables is the major loss-making sub-category with a total loss of 64,083.39.
- Central is the highest-performing region in terms of both sales and profit.
- Phones generated the highest sales among sub-categories.
- Copiers generated the highest profit among sub-categories.
- Higher discount levels generally have a negative impact on profitability.
- Apple Smart Phone, Full Size is the top-selling product.
- Canon imageCLASS 2200 Advanced Copier is the most profitable product.
- Tamara Chand is the highest-profit customer.
